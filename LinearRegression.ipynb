```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota": [2,3,5,4,6]
})

X= df[["horas"]]
y= df[["nota"]]
print (X)

```

```python
#Creamos nuestro modeo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
#Imprimimos el interceptor y el coef
print("Inteceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict(X)
```

```python
# Realizamos la nueva predicción
horas = float(input("ingrese las horas de estudio: "))

# Opción 1: Pasar el dato dentro de un DataFrame con el nombre de la columna (Recomendado)
nuevo_dato = pd.DataFrame({"horas": [horas]})
prediccion = modelo.predict(nuevo_dato)

print("predicción:", prediccion)
```

```python
# Los puntos en azul son los reales, el verde es el dato predicho
plt.figure(figsize=(10, 5))

# 1. Puntos con los datos originales (azul)
plt.scatter(X, y, color='blue', label='Datos reales', s=60)

# 2. Línea de tendencia generada por la regresión lineal
plt.plot(X, y_pred, color='red', linestyle='--', label='Línea de regresión')

# 3. Nuevo punto ingresado y predicho (verde)
plt.scatter(horas, prediccion, color='green', s=120, zorder=5, label='Predicción')

# Etiquetas y formato
plt.xlabel('Horas de estudio')
plt.ylabel('Nota')
plt.title('Regresión Lineal: Horas vs Nota')
plt.legend()
plt.grid(True)

plt.show()
```
